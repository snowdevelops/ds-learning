## 1. The raw data — a dict of lists

A DataFrame usually starts life as plain Python. Here each **key** becomes a
column name, and each **list** becomes that column's values.

Note the `:` separating key from value. `=` is assignment (a statement);
a dict holds key–value *pairs*, so it needs `:`.

```python
data = {
    'name': ['Ana', 'Bruno', 'Carla', 'Diego'],
    'age':  [23, 35, 29, 41],
    'city': ['Uberlandia', 'Sao Paulo', 'Uberlandia', 'Belo Horizonte'],
}
data
```

## 2. dict -> DataFrame

`pd.DataFrame(data)` turns those parallel lists into a table. The lists must all
be the same length — they become rows lined up side by side.

The numbers on the left (0,1,2,3) are the **index** — pandas' row labels.

```python
df = pd.DataFrame(data)
df
```

## 3. Look before you leap

Four methods you will use on literally every dataset you ever touch.

```python
print(df.shape)      # (rows, columns)
print()
print(df.dtypes)     # what type is each column?
```

```python
df.head(2)           # first N rows - vital when you have 2 million
```

```python
df.info()            # columns, non-null counts, memory
```

```python
df.describe()        # stats for NUMERIC columns only
```

## 4. Selecting

- `df['age']` -> one column, a **Series** (1-D)
- `df[['name','age']]` -> several columns, a **DataFrame** (2-D)

The double brackets confuse everyone at first: the inner `[]` is just a *list*
of column names you're handing to the outer `[]`.

```python
df['age']
```

```python
df[['name', 'age']]
```

## 5. Filtering (boolean masking)

This is the single most important pandas idea. Two steps:

1. A comparison produces a column of True/False — a **mask**.
2. Passing that mask to `df[...]` keeps only the True rows.

```python
mask = df['age'] > 30
mask
```

```python
df[mask]
```

```python
# usually written in one line
df[df['city'] == 'Uberlandia']
```

## 6. New columns

```python
df['age_in_10'] = df['age'] + 10
df
```

## 7. Grouping — split, apply, combine

`groupby` splits rows into buckets, applies a calculation to each, and combines
the results. This is the workhorse of data analysis.

```python
df.groupby('city')['age'].mean()
```

```python
df.groupby('city').agg(
    people=('name', 'count'),
    avg_age=('age', 'mean'),
)
```
