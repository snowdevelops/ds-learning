```python
import pandas as pd
```

```python
data = {
    'name' : ['Ana', 'Bruno', 'Carla', 'Diego'],
    'age' : [23, 35, 29, 41],
    'city' : ['Uberlandia', 'Sao Paulo', 'Uberlandia', 'Belo Horizonte']
}
```

```python
df = pd.DataFrame(data)
```

```python
df['job'] = ['Engineer', 'Analyst', 'Engineer', 'Analyst']
```

```python
df.groupby('city')['age'].agg(['mean', 'min', 'max'])
```

```python
df.groupby('city')['age'].mean().sort_values(ascending=False).head(1)
```

```python
salaries = {
    'name': ['Ana', 'Bruno', 'Carla', 'Diego'],
    'salary': [4500, 6200, 5100, 7000]
}
df_salaries = pd.DataFrame(salaries)
```

```python
extra_salaries = {
    'name': ['Ana', 'Bruno', 'Carla'],
    'salary': [4500, 6200, 5100]
}
df_extra = pd.DataFrame(extra_salaries)
result_outer = df.merge(df_extra, on='name', how ='outer')
result_outer
```

```python
extra_salaries2 = {
    'name': ['Ana', 'Bruno', 'Carla', 'Eduardo'],
    'salary': [4500, 6200, 5100,8000]
}
df_extra2 = pd.DataFrame (extra_salaries2)
result_left2 = df.merge(df_extra2, on='name', how='left')
result_left2['salary'] = result_left2['salary'].fillna(result_left2['salary'].mean())
result_left2
```

```python
merged = df.merge(df_salaries, on='name')
merged
```
