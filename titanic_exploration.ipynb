```python
import pandas as pd
```

##loading in the raw titanic file

```python
df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv')
df
```

## Using query to filter only the needed columns

```python
min_age = 18
target_class = None
target_sex = None     

conditions = ["age < @min_age"]
if target_class is not None:
    conditions.append("class == @target_class")

if target_sex is not None:
    conditions.append("sex == @target_sex")

df.query(" and ".join(conditions)) if conditions else df
```

# Filtering by age, class and sex
